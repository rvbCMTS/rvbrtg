```python
import pydicom
from pathlib import Path
from typing import Dict

def compare_dicom_fields(path_file1: Path, path_file2: Path)-> Dict:
    ""
    d1 = pydicom.dcmread(path_file1)
    d2 = pydicom.dcmread(path_file2)

    # Alla taggar i respektive fil
    tags1 = {elem.tag for elem in d1}
    tags2 = {elem.tag for elem in d2}

    # Taggar som finns i båda
    common_tags = tags1 & tags2

    same_fields = []
    different_fields = []

    for tag in common_tags:
        if d1[tag].value == d2[tag].value:
            same_fields.append((tag, d1[tag].name))
        else:
            different_fields.append((tag, d1[tag].name))

    # Taggar som finns i endast ena filen
    unique_tags = tags1 ^ tags2
    unique_fields = []

    for tag in unique_tags:
        if tag in d1:
            unique_fields.append((tag, d1[tag].name))
        else:
            unique_fields.append((tag, d2[tag].name))

    return {
        "same": same_fields,
        "different": different_fields,
        "unique": unique_fields
    }

# användning:

# paths till dom två dicom filerna som ska jämforas
# file_1 = Path(r"")
# file_2 = Path(r"")

# fields = compare_dicom_fields(file_1, file_2)

# printa ut fields["same"], fields["different"] och fields["unique"] for att se var DICOM filerna skiljer sig.
# Funktion framtagen i samband med felsokning av skick mellan capture dator och PACS.

# print(f'\nfält som är samma i båda dcm filerna:\n')
# [print(item) for item in fields["same"]]
# print(f'\nfält som är olika i båda dcm filerna:\n')
# [print(item) for item in fields["different"]]
# print(f'\nfält som är unika for någon av dcm filerna:\n')
# [print(item) for item in fields["unique"]]
```
