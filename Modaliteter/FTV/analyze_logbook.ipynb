```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
import pandas as pd
pd.set_option("display.max_rows", 40)

import plotly.graph_objects as go
from cbct_tools import (
    plot_statistic,
    parse_gender_from_swedish_id,
    parse_datetime,
    parse_patient_age_from_id,
    parse_fov_and_imaging_from_fov_column,
    append_mas,
    parse_scan_column,
    remove_unnecessary_columns,
    translate_column_names_swe_eng,
    plot_summary_statistics
)
```

```python
dp = Path("")

df = pd.read_excel(dp)
df = remove_unnecessary_columns(df)
df = translate_column_names_swe_eng(df)
df = parse_datetime(df, sort_by_datetime=True)
df = parse_patient_age_from_id(df)

df = df[df['Modality'] == 'CT']

df = parse_fov_and_imaging_from_fov_column(df)
df = append_mas(df)
df = parse_scan_column(df)
df = parse_gender_from_swedish_id(df)

# reorder
df = df[
    ['DateTime', 'Modality', 'kV', 'mA', 'Scan time', 'Imaging', 'Scan', 'FOV',
       'DAP (mGycm2)', 'mAs', 'Model Name', 'Patient age', 'Patient Sex']
]

# select machine
df = df[df["Model Name"] == '3D Accuitomo 170']
# df = df[df["Model Name"] == 'Veraview X800']

df

```

```python
# plot_statistics(df, parameter='DAP (mGycm2)', export_to_browser=True)


DSN_1_kvadrant = np.array([0.170, 0.174, 0.177, 0.179, 0.180, 0.187, 0.200, 0.200, 0.210, 0.298, 0.333, 0.350]) # Accuitomo 3D F170 i dosreg
DSN_4_kvadranter = np.array([0.579, 0.580, 0.600, 0.760, 0.782, 0.832, 0.834, 0.963, 1.020, 1.085, 1.101])   # Accuitomo 3D F170 i dosreg

plot_statistic(
    df,
    parameter="DAP (mGycm2)",
    shade_y_regions=[
        (1000 * DSN_1_kvadrant.min(), 1000 * DSN_1_kvadrant.max()),
        (1000 * DSN_4_kvadranter.min(), 1000 * DSN_4_kvadranter.max())
    ],
    shade_labels=[
        "DSN_1_kvadrant",
        "DSN_4_kvadranter"
    ],
    shade_fillcolors=[
        "rgba(0,0,256,0.50)",
        "rgba(256,0,0,0.50)"
    ],
    reference_levels=(200, 700),
    reference_labels=('DRN 1 kvadrant 200 mGycm²', 'DRN 4 kvadrant 700 mGycm²'),
    reference_colors=("rgba(0,0,256,1)", "rgba(256,0,0,1)"),
    export_to_browser=True,
    dark_mode=True,
)

```
